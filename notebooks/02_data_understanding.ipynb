# 2. Data Understanding

## 1. Introduction

Data understanding is an essential stage of the Machine Learning
workflow.

The objective of this stage is to inspect the dataset, understand
its structure, identify the available features, and examine the
target variable.

No data transformation or model training is performed in this stage.

```python
import numpy as np
import pandas as pd
from IPython.display import display

from src.config import (
    EXPECTED_COLUMNS,
    RAW_DATA_PATH,
    TARGET_COLUMN,
)
from src.data_loader import load_data

df = load_data(RAW_DATA_PATH)

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)

display(df.head())
```

## 2. Dataset Structure

The dataset contains transaction-related information.

The expected columns are:

- Time: Time elapsed since the first transaction in the dataset.
- V1 to V28: Anonymized numerical features.
- Amount: Transaction amount.
- Class: Target variable indicating whether a transaction is fraudulent.

Target variable:

- 0 represents a legitimate transaction.
- 1 represents a fraudulent transaction.

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumn Names:")

for index, column in enumerate(df.columns, start=1):
    print(f"{index}. {column}")
```

```python
print("Data Types:")

display(
    df.dtypes.to_frame(name="Data Type")
)

print("\nDataset Information:")

df.info()
```

```python
print("Descriptive Statistics:")

display(
    df.describe().T
)
```

## 3. Target Variable Analysis

The target variable is `Class`.

This is a binary classification target.

The distribution of the target classes must be investigated
before selecting a suitable training strategy.

A significant difference between the number of legitimate
and fraudulent transactions may indicate class imbalance.

```python
target_counts = df[TARGET_COLUMN].value_counts().sort_index()

target_percentages = (
    df[TARGET_COLUMN]
    .value_counts(normalize=True)
    .sort_index() * 100
)

target_summary = pd.DataFrame({
    "Count": target_counts,
    "Percentage": target_percentages
})

display(target_summary)

print("\nClass 0: Legitimate Transactions")
print("Class 1: Fraudulent Transactions")
```

```python
total_transactions = len(df)

fraud_count = int(
    (df[TARGET_COLUMN] == 1).sum()
)

legitimate_count = int(
    (df[TARGET_COLUMN] == 0).sum()
)

fraud_percentage = (
    fraud_count / total_transactions
) * 100

legitimate_percentage = (
    legitimate_count / total_transactions
) * 100

print("Total Transactions:", total_transactions)
print("Legitimate Transactions:", legitimate_count)
print("Fraudulent Transactions:", fraud_count)

print(f"\nFraud Percentage: {fraud_percentage:.4f}%")
print(f"Legitimate Percentage: {legitimate_percentage:.4f}%")
```

## 4. Missing Values Analysis

Missing values may affect the quality of the dataset.

In this section, the number and percentage of missing values
are calculated for every column.

Missing values will be handled in a later stage using
a preprocessing pipeline fitted only on training data.

```python
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().mean() * 100
    )
})

display(
    missing_summary.sort_values(
        by="Missing Percentage",
        ascending=False
    )
)
```

## 5. Duplicate Records Analysis

Duplicate records may indicate repeated observations.

However, duplicate transactions should not automatically be
considered invalid.

The duplicate count will be investigated before deciding
whether any records should be removed.

```python
duplicate_count = df.duplicated().sum()

duplicate_percentage = (
    duplicate_count / len(df)
) * 100

print("Duplicate Rows:", duplicate_count)

print(
    f"Duplicate Percentage: {duplicate_percentage:.4f}%"
)

if duplicate_count > 0:
    display(
        df[df.duplicated(keep=False)].head(10)
    )
```
