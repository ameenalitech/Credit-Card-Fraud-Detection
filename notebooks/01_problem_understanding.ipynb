# Credit Card Fraud Detection

## 1. Project Overview

Credit card fraud is a major challenge in the financial industry.
Fraudulent transactions can cause significant financial losses
for customers, banks, and financial institutions.

The main objective of this project is to develop a Machine Learning
system capable of identifying potentially fraudulent credit card
transactions.

The project follows a complete Machine Learning workflow, including:

## 2. Business Problem

Financial institutions process a large number of credit card
transactions every day.

A small proportion of these transactions may be fraudulent.

The business challenge is to identify fraudulent transactions
while minimizing two major types of errors:

1. False Positive:
   A legitimate transaction is incorrectly classified as fraud.

2. False Negative:
   A fraudulent transaction is incorrectly classified as legitimate.

False negatives may result in financial losses, while false positives
may negatively affect customer experience.

Therefore, the model must achieve a suitable balance between
fraud detection performance and false alarm rates.

## 3. Machine Learning Problem Definition

This project is a Supervised Binary Classification problem.

The target variable is:

- Class = 0: Legitimate transaction
- Class = 1: Fraudulent transaction

The model receives transaction-related features as input
and predicts the probability that a transaction is fraudulent.

The final prediction is generated using a classification threshold.

## 5. Evaluation Metrics

Accuracy alone is not sufficient for evaluating fraud detection
models because the dataset is expected to be highly imbalanced.

The following metrics will be considered:

- Precision
- Recall
- F1-Score
- ROC-AUC
- Average Precision
- Confusion Matrix
- Matthews Correlation Coefficient (MCC)
- Balanced Accuracy

Recall is particularly important because it measures the proportion
of actual fraudulent transactions correctly detected.

Precision is also important because it measures how many transactions
predicted as fraud are actually fraudulent.

Average Precision will be used as an important model-selection
metric during cross-validation.
